```python
# A network security team wants to detect unusual system activity.
# Build an LSTM autoencoder to reconstruct normal patterns and explain
# how reconstruction error helps identify anomalies.

import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import classification_report

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, RepeatVector, TimeDistributed, Dense


# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)
```

```python
# Step 2: lstm cell class
class LSTMCell:
    def __init__(self, input_size, hidden_size):
        self.input_size = input_size
        self.hidden_size = hidden_size
```

```python
class LSTMCell:
    def __init__(self, input_size, hidden_size):
        self.input_size = input_size
        self.hidden_size = hidden_size

        # Weights for gates
        self.Wf = np.random.randn(hidden_size, input_size + hidden_size)
        self.Wi = np.random.randn(hidden_size, input_size + hidden_size)
        self.Wc = np.random.randn(hidden_size, input_size + hidden_size)
        self.Wo = np.random.randn(hidden_size, input_size + hidden_size)

        # Biases
        self.bf = np.zeros((hidden_size, 1))
        self.bi = np.zeros((hidden_size, 1))
        self.bc = np.zeros((hidden_size, 1))
        self.bo = np.zeros((hidden_size, 1))

    def forward(self, x, h_prev, c_prev):
        # Concatenate h_prev & input x
        combined = np.vstack((x, h_prev))

        # FORGET GATE
        f_t = sigmoid(np.dot(self.Wf, combined) + self.bf)

        # INPUT GATE
        i_t = sigmoid(np.dot(self.Wi, combined) + self.bi)
        c_s = tanh(np.dot(self.Wc, combined) + self.bc)

        # Cell state update
        c_t = f_t * c_prev + i_t * c_s

        # Output gate
        o_t = sigmoid(np.dot(self.Wo, combined) + self.bo)

        # Hidden state update
        h_t = o_t * tanh(c_t)

        return h_t, c_t, f_t, i_t, o_t

np.random.seed(1)
input_size = 1
hidden_size = 3
sequence_length = 4

# Sample input sequence
X = [np.random.randn(input_size, 1) for _ in range(sequence_length)]

# Initialization
lstm = LSTMCell(input_size, hidden_size)

# Initial states
h = np.zeros((hidden_size, 1))
c = np.zeros((hidden_size, 1))

# Forward pass through sequence
print("Step by step LSTM EXECUTION:\n")
for t in range(sequence_length):
    x = X[t]
    h, c, f, i, o = lstm.forward(x, h, c)

    print(f"Time step {t+1}")
    print("Input :", x.flatten())
    print("Hidden State :", h.flatten())
    print("Forget Gate :", f.flatten())
    print("Input Gate :", i.flatten())
    print("Output Gate :", o.flatten())
    print("Cell State :", c.flatten())
    print("-" * 40)
```
